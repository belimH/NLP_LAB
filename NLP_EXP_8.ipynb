```python
#Hamzah Belim
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
```

```python
#Hamzah Belim
documents = [
    "Artificial intelligence allows machines to solve complex problems.",
    "Machines can solve complex problems using artificial intelligence.",
    "Computers can process and understand human language.",
    "The students visited the museum during their college trip.",
    "The weather was pleasant throughout the morning."
]
```

```python
#Hamzah Belim
for i, document in enumerate(documents):
    print(f"D{i + 1}: {document}")
```

```python
#Hamzah Belim
vectorizer = TfidfVectorizer(stop_words='english')

tfidf_matrix = vectorizer.fit_transform(documents)

print(tfidf_matrix.shape)
```

```python
#Hamzah Belim
tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=vectorizer.get_feature_names_out(),
    index=["D1", "D2", "D3", "D4", "D5"]
)

tfidf_df.round(2)
```

```python
#Hamzah Belim
print(vectorizer.get_feature_names_out())
```

```python
#Hamzah Belim
similarity_matrix = cosine_similarity(tfidf_matrix)

similarity_matrix
```

```python
#Hamzah Belim
similarity_df = pd.DataFrame(
    similarity_matrix,
    index=["D1", "D2", "D3", "D4", "D5"],
    columns=["D1", "D2", "D3", "D4", "D5"]
)

similarity_df.round(3)
```

```python
#Hamzah Belim
highest_similarity = -1
most_similar_pair = None

for i in range(len(documents)):
    for j in range(i + 1, len(documents)):
        similarity = similarity_matrix[i][j]

        if similarity > highest_similarity:
            highest_similarity = similarity
            most_similar_pair = (i, j)

print(
    f"Most similar documents: D{most_similar_pair[0] + 1} "
    f"and D{most_similar_pair[1] + 1}"
)

print(f"Similarity: {highest_similarity:.3f}")
```
